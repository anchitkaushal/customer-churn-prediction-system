```python

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df = pd.read_csv("../../data/raw/customer churn.csv")
```

```python
new_df = df.head(500)
```

```python
for i , predictor in enumerate(new_df.drop(columns=['RowNumber','CustomerId','Surname','Geography','Gender','Card Type',"NumOfProducts","Credit Card","IsActiveMember","Complain","Satisfaction Score"])): 
    plt.figure(i,figsize=(10,5))
    sns.boxplot(data = new_df, x = predictor ,hue = 'Churned' )
    plt.title(f"{predictor} vs {"Churned"}")
```

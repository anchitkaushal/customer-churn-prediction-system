```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df = pd.read_csv("../../data/raw/customer churn.csv")
```

```python
new_df = df.head(500)
```

```python
def plot_num_num(df, x, y):
    plt.figure(figsize=(6, 5))

    sns.scatterplot(
        data=df,
        x=x,
        y=y,
        
    )

    plt.title(f"{x} vs {y}")
    plt.show()
```

```python
combinations = [
('Balance' ,'EstimatedSalary'),
('Age','Balance'),
('Age','CreditScore'),
('Tenure','Balance'),
('Point Earned','Balance'),
('Tenure','Point Earned'),
]
for x , y in combinations:
   plot_num_num(df, x, y)
 
```

```python
continuous_features = ["Balance","EstimatedSalary","CreditScore","Age","Point Earned" , "Tenure"]
```

```python
corr_matrix = new_df[continuous_features].corr()
plt.figure(figsize = (10,5))
sns.heatmap(
    corr_matrix,
    annot=True,        # Show correlation values
    cmap=sns.cubehelix_palette(as_cmap=True),# Color scheme
    fmt=".2f",         # Format numbers
    linewidths=0.5   # Color scheme
    )
```

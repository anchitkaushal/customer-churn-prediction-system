```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df = pd.read_csv("../../data/raw/customer churn.csv")
```

```python
df.shape
```

```python
for i , predictor in enumerate(df.drop(columns=['RowNumber','CustomerId','Surname','Balance','EstimatedSalary','CreditScore','Age','Point Earned','Tenure'])): 
    plt.figure(i,figsize=(10,5))
    sns.countplot(data = df, x = predictor , hue = "Churned" )
    plt.title(f"{predictor} vs {"Churned"}")
    plt.show()
```

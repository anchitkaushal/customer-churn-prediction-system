```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../../data/raw/customer churn.csv")
```

```python
df.columns

```

```python
new_df = df.head(500)
```

Analysis for categorcial features


```python
for i , predictor in enumerate(new_df.drop(columns=['RowNumber','CustomerId','Surname','Geography','Gender','Card Type','NumOfProducts','Credit Card',
       'IsActiveMember', 'Churned', 'Complain','Satisfaction Score', 'Card Type'])):
    plt.figure(i,figsize=(10,5))
    sns.boxplot(data = new_df, x = predictor)
```

```python
for i , predictor in enumerate(new_df.drop(columns=['RowNumber','CustomerId','Surname','Geography','Gender','Card Type','NumOfProducts','Credit Card',
       'IsActiveMember', 'Churned', 'Complain','Satisfaction Score', 'Card Type'])):
    plt.figure(i,figsize=(10,5))
    sns.histplot(data = new_df[predictor] , bins= 5 , kde= True)
```

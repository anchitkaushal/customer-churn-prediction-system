```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df = pd.read_csv("../../data/raw/customer churn.csv")
```

```python
new_df = df.head(500)
```

```python
pd.set_option('display.max_columns',None)
dmydf = pd.get_dummies(new_df.drop(columns= ['Surname','RowNumber','CustomerId']),dtype= int)
dmydf.head()
```

```python
plt.figure(figsize = (20,8))
dmydf.corr()['Churned'].sort_values(ascending = False).plot(kind ='bar')
```

```python
corr_matrix = dmydf.corr()
```

```python
plt.figure(figsize = (13,7))
sns.heatmap(
    corr_matrix,
    annot=True,        # Show correlation values
    cmap=sns.cubehelix_palette(as_cmap=True),   # Color scheme
    fmt=".2f",         # Format numbers
    linewidths=0.5 )
```
